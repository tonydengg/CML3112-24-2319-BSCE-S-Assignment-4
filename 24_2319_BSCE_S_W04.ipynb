```python
STUDENT_ID="24/2319/BSCE-S"
COHORT="CML3112"
import csv
with open("/content/site_inspection_log.csv", newline="") as f:
    rows = list(csv.DictReader(f))
```

## valid_count (5 marks)
Store an integer count of rows whose crack_width_mm field is nonblank. Skip blank readings before numeric conversion.


```python
valid_count = 0
for row in rows:
    if row.get('crack_width_mm', '').strip() != '':
        valid_count += 1
print(f"Valid count: {valid_count}")
```

## mean_reading (5 marks)
Store the arithmetic mean of the nonblank crack_width_mm values as a number. Add the valid readings and divide by their count. Keep full precision.


```python
total_width = 0.0
for row in rows:
    val = row.get('crack_width_mm', '').strip()
    if val != '':
        total_width += float(val)

mean_reading = total_width / valid_count if valid_count > 0 else 0.0
print(f"Mean reading: {mean_reading} mm")
```

## above_limit (5 marks)
Store an integer count of valid crack_width_mm readings strictly greater than 0.3. A reading equal to the threshold is not counted.


```python
above_limit = 0
for row in rows:
    val = row.get('crack_width_mm', '').strip()
    if val != '':
        if float(val) > 0.3:
            above_limit += 1
print(f"Above limit (> 0.3 mm): {above_limit}")
```

## group_means (5 marks)
Store a dictionary whose keys are the structure_id labels and whose values are the mean valid crack_width_mm readings within each group. Divide each group total by its own valid count. Keep full precision.


```python
group_totals = {}
group_counts = {}
for row in rows:
    val = row.get('crack_width_mm', '').strip()
    struct_id = row.get('structure_id', '').strip()
    if val != '' and struct_id != '':
        val_float = float(val)
        group_totals[struct_id] = group_totals.get(struct_id, 0.0) + val_float
        group_counts[struct_id] = group_counts.get(struct_id, 0) + 1

group_means = {}
for struct_id in group_totals:
    group_means[struct_id] = group_totals[struct_id] / group_counts[struct_id]

print("Group means:")
for k, v in group_means.items():
    print(f"  {k}: {v} mm")
```
